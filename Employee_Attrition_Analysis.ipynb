# Employee Attrition Analysis

## Objective
The goal of this project is to analyze employee attrition data and identify the factors that influence employees leaving the company.

## Problem Statement
Employee attrition can impact productivity and increase hiring costs. By analyzing employee data, we can identify patterns and help organizations improve employee retention.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(x='Attrition', data=df)
plt.show()
```

```python
sns.countplot(
    x='Department',
    hue='Attrition',
    data=df
)
plt.show()
```

```python
sns.boxplot(
    x='Attrition',
    y='MonthlyIncome',
    data=df
)
plt.show()
```

```python
sns.histplot(
    data=df,
    x='Age',
    hue='Attrition'
)
plt.show()
```

Research & Development department recorded the highest number of employee exits.

Employees with lower monthly income showed higher attrition rates.

Employees aged 26–35 experienced the highest turnover.

# Conclusion

The analysis revealed that employee attrition is influenced by several factors, including department, salary, age, and overtime.

Key Findings:
- Research & Development showed the highest attrition.
- Lower-income employees were more likely to leave.
- Employees aged 26–35 had the highest turnover.
- Overtime employees showed significantly higher attrition.

These insights can help organizations develop strategies to improve employee retention and reduce turnover.
